```python
from collections import Counter
```

```python
def read_data(path):
    edits = []
    all_edits = []

    with open(path) as f:
        for line in f.readlines()[1:]:
            if line.strip() == '':
                all_edits.append(edits)
                edits  = []
            else:
                line = line.replace('\n', '').split('\t')
                assert len(line) == 2
                src, tgt = line
    
                if src == tgt:
                    edit = 'UC'

                elif src == '': # insert
                    assert tgt != ''
                    edit = 'Insert'
                
                elif tgt == '':
                    assert src != '' # deletion
                    edit = 'Delete'
                
                else:
                    assert src != '' and tgt != ''
                    edit = 'Replace'

                edits.append({'src': src, 'tgt': tgt, 'edit': edit})
    
    if edits:
        all_edits.append(edits)
    
    return all_edits
            
```

```python
def cnt_edits(data):
    all_edits  = [token['edit'] for example in data for token in example]
    cnts = Counter(all_edits)
    return {'UC': cnts['UC'], 'Replace' : cnts['Replace'], 'Insert': cnts['Insert'], 'Delete': cnts['Delete']}
```

```python
train = read_data('/scratch/ba63/zaebuc-lrec-2026/expriments/written/gec/data/zaebuc-w1/ar/alignment/train.txt')
dev = read_data('/scratch/ba63/zaebuc-lrec-2026/expriments/written/gec/data/zaebuc-w1/ar/alignment/dev.txt')
test = read_data('/scratch/ba63/zaebuc-lrec-2026/expriments/written/gec/data/zaebuc-w1/ar/alignment/test.txt')
```

```python
train_edits, dev_edits, test_edits = cnt_edits(train), cnt_edits(dev), cnt_edits(test)
print(train_edits)
print(dev_edits)
print(test_edits)
```
